## 1. Introdução

Este notebook apresenta uma aplicação de conceitos de Álgebra Linear em dados utilizados por um agente de atendimento acadêmico. A proposta consiste em transformar informações relacionadas aos alunos em representações matemáticas por meio de vetores e matrizes, permitindo realizar comparações quantitativas entre diferentes perfis.

Foram utilizados dados presentes nos arquivos `Matriculas.xlsx`, `Historico.xlsx`, `Relacionamentos.xlsx` e `Financeiro.xlsx`. A partir dessas fontes, foram selecionadas características acadêmicas, de relacionamento e financeiras que contextualizam uma solicitação feita ao agente.

## 2. Dados utilizados pelo agente

Para representar cada aluno, foram selecionadas cinco características numéricas. Como as informações possuem escalas diferentes, os valores foram normalizados para facilitar a comparação matemática.

| Característica | Origem | Representação |
|---|---|---|
| Progresso acadêmico | Matriculas.xlsx | Período atual / 8 |
| Nota | Historico.xlsx | Média / 100 |
| Assiduidade | Historico.xlsx | Média / 100 |
| Relacionamentos | Relacionamentos.xlsx | Contatos / 143 |
| Regularidade financeira | Financeiro.xlsx | Lançamentos baixados / total |

A normalização é importante porque, sem ela, uma característica com valores numericamente maiores poderia ter influência desproporcional nas operações de distância e similaridade.

## 3. Representação dos dados em vetores

Cada aluno foi representado por um vetor de cinco dimensões, seguindo a ordem:

**v = [progresso, nota, assiduidade, contatos, regularidade financeira]**

Foram utilizados dois perfis para demonstrar as operações:

| Aluno | Progresso | Nota | Assiduidade | Contatos | Regularidade |
|---|---|---|---|---|---|
| 12621 | 0,8750 | 0,6240 | 1,0000 | 0,1049 | 0,2500 |
| 17578 | 1,0000 | 0,8209 | 1,0000 | 0,0000 | 1,0000 |

```python
import numpy as np

# Vetor do aluno 12621
aluno_12621 = np.array([
    7/8,          # progresso acadêmico
    62.4/100,     # nota média
    100/100,      # assiduidade
    15/143,       # contatos / relacionamentos
    0.25          # regularidade financeira
])

# Vetor do aluno 17578
aluno_17578 = np.array([
    8/8,
    82.088947/100,
    100/100,
    0/143,
    1.00
])

print("v1 (aluno 12621):", aluno_12621)
print("v2 (aluno 17578):", aluno_17578)
```

## 4. Representação em matriz

Os dois vetores podem ser organizados em uma matriz, na qual cada linha representa um aluno e cada coluna representa uma característica.

```python
M = np.array([aluno_12621, aluno_17578])
print("Matriz M (linhas = alunos, colunas = características):")
print(M)
print("Formato (shape):", M.shape)
```

## 5. Operações de Álgebra Linear

Foram realizadas quatro operações: **norma**, **produto escalar**, **distância euclidiana** e **ângulo entre vetores**.

### 5.1 Norma dos vetores

A norma euclidiana mede o tamanho de um vetor:

$$\|v\| = \sqrt{v_1^2 + v_2^2 + \dots + v_n^2}$$

A norma representa a magnitude matemática do vetor e não deve ser interpretada isoladamente como uma classificação de desempenho, pois depende das características selecionadas.

```python
norma_12621 = np.linalg.norm(aluno_12621)
norma_17578 = np.linalg.norm(aluno_17578)

print(f"Norma do aluno 12621: {norma_12621:.4f}")
print(f"Norma do aluno 17578: {norma_17578:.4f}")
```

### 5.2 Produto escalar

O produto escalar combina os componentes correspondentes de dois vetores:

$$u \cdot v = u_1v_1 + u_2v_2 + \dots + u_nv_n$$

Como as características estão normalizadas, o produto escalar pode ser utilizado como parte de uma medida de proximidade ou similaridade entre perfis.

```python
produto_escalar = np.dot(aluno_12621, aluno_17578)
print(f"Produto escalar v1 · v2: {produto_escalar:.4f}")
```

### 5.3 Distância euclidiana

A distância euclidiana mede o afastamento entre dois vetores:

$$d(u,v) = \|u - v\|$$

Uma distância menor indica que os perfis são mais próximos considerando as características escolhidas. Essa operação pode apoiar o agente na identificação de casos ou alunos com características semelhantes.

```python
distancia = np.linalg.norm(aluno_12621 - aluno_17578)
print(f"Distância euclidiana d(v1, v2): {distancia:.4f}")
```

### 5.4 Ângulo entre os vetores

O ângulo entre dois vetores pode ser obtido a partir do produto escalar:

$$\theta = \arccos\left(\frac{u \cdot v}{\|u\|\,\|v\|}\right)$$

Um ângulo menor indica que os vetores possuem direções mais semelhantes no espaço das características, fornecendo outra forma de analisar a proximidade entre os perfis.

```python
cosseno = produto_escalar / (norma_12621 * norma_17578)
angulo_graus = np.degrees(np.arccos(np.clip(cosseno, -1, 1)))

print(f"Cosseno do ângulo: {cosseno:.4f}")
print(f"Ângulo entre v1 e v2: {angulo_graus:.2f}°")
```

## 6. Resumo dos resultados

```python
print(f"{'Operação':<30}{'Resultado'}")
print("-" * 45)
print(f"{'Norma — aluno 12621':<30}{norma_12621:.4f}")
print(f"{'Norma — aluno 17578':<30}{norma_17578:.4f}")
print(f"{'Produto escalar':<30}{produto_escalar:.4f}")
print(f"{'Distância euclidiana':<30}{distancia:.4f}")
print(f"{'Ângulo entre os vetores':<30}{angulo_graus:.2f}°")
```

## 7. Interpretação para o funcionamento do agente

A aplicação da Álgebra Linear permite transformar diferentes tipos de informação do sistema acadêmico em uma representação numérica única. Em vez de analisar cada informação de maneira isolada, o agente pode trabalhar com um vetor que representa o perfil do aluno.

Por exemplo, quando um estudante realiza uma solicitação, seu perfil pode ser transformado em um vetor e comparado com outros perfis ou casos históricos:

- A **norma** indica a magnitude geral do perfil dentro do espaço de características escolhido.
- O **produto escalar** funciona como uma medida bruta de similaridade entre dois perfis.
- A **distância euclidiana** (≈ 0,7924) mostra o quão afastados estão os dois perfis analisados — quanto menor, mais parecidos.
- O **ângulo** (≈ 22,83°) mostra que, apesar de os valores absolutos serem diferentes, a *direção* dos dois perfis no espaço de características é relativamente semelhante.

É importante destacar que essas operações não determinam sozinhas a resposta que o agente deve fornecer. Elas funcionam como ferramentas matemáticas de comparação e representação, que podem complementar mecanismos de busca, regras de negócio e modelos de Inteligência Artificial, auxiliando o agente a contextualizar solicitações e recuperar informações relevantes.
